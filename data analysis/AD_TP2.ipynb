```python
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
import seaborn as sns

# Création du DataFrame
data = pd.DataFrame({
    'Temperature': [14.2, 16.4, 11.9, 15.2, 18.5, 22.1,
                    19.4, 25.1, 23.4, 18.1, 22.6, 17.2],
    'Sales': [215.2, 325.0, 185.9, 330.1, 460.1, 520.0,
              450.2, 610.1, 490.5, 410.2, 470.2, 360.5]
})

# Nuage de points
sns.scatterplot(x='Temperature', y='Sales', data=data)
plt.title("Ventes vs Température")
plt.show()

# Variables X et Y
X = sm.add_constant(data[['Temperature']])  # مهم: DataFrame ماشي Series
Y = data['Sales']

# Estimation du modèle
modele_reg = sm.OLS(Y, X).fit()

# Résumé
print(modele_reg.summary())

# Prédiction pour 20°C
temp_pred = pd.DataFrame({
    'const': [1],
    'Temperature': [20]
})

prediction = modele_reg.predict(temp_pred)

print(f"\nPrédiction pour 20 °C : {prediction.iloc[0]:.2f} k unités")
```
